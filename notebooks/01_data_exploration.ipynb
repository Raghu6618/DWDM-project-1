# 01. Exploratory Data Analysis (EDA)
## Customer Segmentation in Banking DWDM Project
This notebook inspects the raw dataset ank_transactions.csv, calculates summary statistics, checks missingness, data types, distributions, and generates initial visualizations.

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load raw transaction dataset
df = pd.read_csv('../data/raw/bank_transactions.csv')
print(f'Total Rows: {len(df):,}')
print(f'Total Columns: {len(df.columns)}')
print(df.info())
```

```
# Column summary and missing values
missing = df.isnull().sum()
pct_missing = (missing / len(df)) * 100
summary_df = pd.DataFrame({'Missing_Count': missing, 'Missing_Pct': pct_missing.round(2)})
summary_df
```

```
# Descriptive statistics for numerical variables
df[['CustAccountBalance', 'TransactionAmount (INR)', 'TransactionTime']].describe()
```

```
# Unique customer counts and transaction frequency per customer
tx_per_cust = df['CustomerID'].value_counts()
print(f'Unique Customers: {len(tx_per_cust):,}')
print('Transaction count distribution per customer:')
print(tx_per_cust.value_counts(normalize=True).head(5) * 100)
```
