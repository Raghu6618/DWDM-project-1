# 04. Unsupervised Clustering & Model Evaluation
Covers feature standardization, log transformation, Elbow Method, Silhouette Analysis, and K-Means fitting.

```
import pandas as pd
import matplotlib.pyplot as plt

# Load precomputed K evaluation metrics
k_eval = pd.read_csv('../outputs/tables/k_evaluation_metrics.csv')
k_eval
```

```
# Plot Elbow Method and Silhouette Curve
fig, ax1 = plt.subplots(figsize=(8, 4))
ax1.plot(k_eval['K'], k_eval['Inertia'], 'b-o', label='Inertia (Elbow)')
ax1.set_xlabel('K (Number of Clusters)')
ax1.set_ylabel('Inertia', color='b')
ax2 = ax1.twinx()
ax2.plot(k_eval['K'], k_eval['Silhouette_Score'], 'g-s', label='Silhouette Score')
ax2.set_ylabel('Silhouette Score', color='g')
plt.title('K-Means Evaluation: Elbow Curve & Silhouette Scores')
plt.show()
```
